```python
import matplotlib.pyplot as plt
import numpy as np

# Intervalo temporal discreto
n = np.arange(-5, 11)

# 1. Impulso Unitário delta[n]
delta = np.where(n == 0, 1, 0)

# 2. Degrau Unitário u[n]
u = np.where(n >= 0, 1, 0)

# 3. Rampa Unitária r[n] = n * u[n]
r = np.where(n >= 0, n, 0)

# 4. Exponenciais Discretas
a1 = 0.8  # Decaimento
a2 = 1.15  # Crescimento
exp_dec = np.where(n >= 0, a1**n, 0)
exp_cres = np.where(n >= 0, a2**n, 0)

# Plotagem das Sequências
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.stem(n, delta, basefmt="k-")
plt.title(r"Impulso Unitário $\delta[n]$")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid(True)

plt.subplot(2, 2, 2)
plt.stem(n, u, basefmt="k-")
plt.title(r"Degrau Unitário $u[n]$")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid(True)

plt.subplot(2, 2, 3)
plt.stem(n, r, basefmt="k-")
plt.title(r"Rampa Unitária $r[n]$")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid(True)

plt.subplot(2, 2, 4)
plt.stem(
    n,
    exp_dec,
    linefmt="b-",
    markerfmt="bo",
    basefmt="k-",
    label=r"$a = 0.8$ (Decrescente)",
)
plt.stem(
    n,
    exp_cres,
    linefmt="r--",
    markerfmt="rs",
    basefmt="k-",
    label=r"$a = 1.15$ (Crescente)",
)
plt.title(r"Exponencial Discreta $a^n u[n]$")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
