```python
import numpy as np
import matplotlib.pyplot as plt

# Definindo o vetor de índices n e a sequência base x[n]
n = np.arange(-5, 6)

def x_func(n_array):
    vals = np.zeros_like(n_array, dtype=float)
    # Define x[n] = {1, 2, 3, 2, 1} para n = 0, 1, 2, 3, 4
    base = {0: 1.0, 1: 2.0, 2: 3.0, 3: 2.0, 4: 1.0}
    for i, nv in enumerate(n_array):
        vals[i] = base.get(nv, 0.0)
    return vals

# Operações investigadas
x_orig = x_func(n)
x_delay = x_func(n - 2)     # x[n - 2] (Atraso)
x_advance = x_func(n + 2)   # x[n + 2] (Avanço)
x_inv = x_func(-n)          # x[-n]    (Inversão Temporal)
x_scale = 2 * x_orig        # 2x[n]    (Amplificação)
x_neg = -x_orig             # -x[n]    (Inversão de Polaridade)

# Plotagem dos gráficos
transformacoes = [
    ("Original: x[n]", x_orig),
    ("Atrasado: x[n - 2]", x_delay),
    ("Avançado: x[n + 2]", x_advance),
    ("Invertido no Tempo: x[-n]", x_inv),
    ("Escalonado: 2x[n]", x_scale),
    ("Invertido na Fase: -x[n]", x_neg)
]

plt.figure(figsize=(12, 8))
for i, (titulo, sinal) in enumerate(transformacoes, 1):
    plt.subplot(3, 2, i)
    plt.stem(n, sinal, basefmt=" ")
    plt.title(titulo)
    plt.xlabel("n")
    plt.ylabel("Amplitude")
    plt.grid(True)

plt.tight_layout()
plt.show()
```
