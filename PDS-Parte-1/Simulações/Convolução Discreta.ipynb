```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definição do sinal de entrada x[n] (Tamanho Nx = 3, n = 0..2)
x = np.array([1, 2, 1])

# 2. Resposta ao impulso original h1[n] (Média móvel / Média local)
h1 = np.array([1, 1])

# 3. Resposta ao impulso modificada h2[n] (Diferenciador / Detector de borda)
h2 = np.array([1, -1])

# Convoluções computacionais
y1 = np.convolve(x, h1)  # Experimento 1
y2 = np.convolve(x, h2)  # Experimento 2 (Modificado)

# Vetores de tempo
n_x = np.arange(0, len(x))
n_h = np.arange(0, len(h1))
n_y = np.arange(0, len(y1))

print("=== RESULTADOS COMPUTACIONAIS ===")
print(f"x[n]  = {x}")
print(f"h1[n] = {h1}  -> y1[n] = {y1} (Tamanho: {len(y1)})")
print(f"h2[n] = {h2} -> y2[n] = {y2} (Tamanho: {len(y2)})")

# Plotagem dos sinais
plt.figure(figsize=(12, 6))

plt.subplot(2, 3, 1); plt.stem(n_x, x, basefmt=" "); plt.title("Entrada x[n]")
plt.subplot(2, 3, 2); plt.stem(n_h, h1, basefmt=" "); plt.title("Filtro h1[n] = {1, 1}")
plt.subplot(2, 3, 3); plt.stem(n_y, y1, basefmt=" "); plt.title("Saída y1[n] = x*h1")

plt.subplot(2, 3, 4); plt.stem(n_x, x, basefmt=" "); plt.title("Entrada x[n]")
plt.subplot(2, 3, 5); plt.stem(n_h, h2, basefmt=" "); plt.title("Filtro h2[n] = {1, -1}")
plt.subplot(2, 3, 6); plt.stem(n_y, y2, basefmt=" "); plt.title("Saída y2[n] = x*h2")

plt.tight_layout()
plt.show()
```
