```python
import matplotlib.pyplot as plt
import numpy as np

# Domínio do tempo discreto
N_amostras = 30
n = np.arange(0, N_amostras)

# 1. Respostas ao Impulso h[n]
h1 = (0.5) ** n  # Estável (|0.5| < 1)
h2 = (1.15) ** n  # Instável (|1.15| > 1)

# Entrada Degrau Unitário x[n] = u[n]
x = np.ones(N_amostras)

# 2. Resposta dos sistemas à entrada degrau via convolução
y1 = np.convolve(x, h1)[:N_amostras]
y2 = np.convolve(x, h2)[:N_amostras]

# 3. Verificação de Estabilidade Sum(|h[n]|)
soma_h1 = np.sum(np.abs(h1))
soma_h2 = np.sum(np.abs(h2))

print(f"Soma Módulos Sistema 1 (Estável): {soma_h1:.4f} (Finito)")
print(f"Soma Módulos Sistema 2 (Instável): {soma_h2:.4f} (Crescente/Divergente)")

# Plotagem dos Gráficos
plt.figure(figsize=(12, 6))

# Subplot 1: Sistema LTI Estável
plt.subplot(2, 2, 1)
plt.stem(n, h1, linefmt="b-", markerfmt="bo", basefmt="k-")
plt.title(r"Resposta ao Impulso Estável: $h_1[n] = (0.5)^n u[n]$")
plt.xlabel("n")
plt.ylabel("h1[n]")
plt.grid(True)

plt.subplot(2, 2, 2)
plt.plot(n, y1, "b-o")
plt.title(r"Resposta ao Degrau $y_1[n]$ (Limitada / Convergente)")
plt.xlabel("n")
plt.ylabel("y1[n]")
plt.grid(True)

# Subplot 2: Sistema LTI Instável
plt.subplot(2, 2, 3)
plt.stem(n, h2, linefmt="r-", markerfmt="ro", basefmt="k-")
plt.title(r"Resposta ao Impulso Instável: $h_2[n] = (1.15)^n u[n]$")
plt.xlabel("n")
plt.ylabel("h2[n]")
plt.grid(True)

plt.subplot(2, 2, 4)
plt.plot(n, y2, "r-s")
plt.title(r"Resposta ao Degrau $y_2[n]$ (Ilimitada / Divergente)")
plt.xlabel("n")
plt.ylabel("y2[n]")
plt.grid(True)

plt.tight_layout()
plt.show()
```
