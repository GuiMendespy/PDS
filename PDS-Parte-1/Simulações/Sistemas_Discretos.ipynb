```python
import numpy as np

# Vetor temporal discreto
n = np.arange(-2, 6)

# Entrada limitada: Degrau unitário x[n] = u[n]
x = np.where(n >= 0, 1.0, 0.0)

# Implementação dos Sistemas Discretos
y1 = 2 * x                             # S1: y[n] = 2x[n]
y2 = x + np.roll(x, 1)                 # S2: y[n] = x[n] + x[n-1]
y2[n < -1] = 0                         # Ajuste de borda para roll
y3 = x**2                              # S3: y[n] = x^2[n]
y4 = np.roll(x, -1)                    # S4: y[n] = x[n+1]
y4[n == 5] = 0                         # Ajuste de borda
y5 = n * x                             # S5: y[n] = n*x[n]

# Exibição dos resultados numéricos no console
print(f"n      : {list(n)}")
print(f"x[n]   : {list(x.astype(int))}")
print(f"y1[n]  : {list(y1.astype(int))}")
print(f"y2[n]  : {list(y2.astype(int))}")
print(f"y3[n]  : {list(y3.astype(int))}")
print(f"y4[n]  : {list(y4.astype(int))}")
print(f"y5[n]  : {list(y5.astype(int))}")
```
