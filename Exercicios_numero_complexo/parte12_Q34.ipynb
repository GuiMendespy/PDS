```python
import matplotlib.pyplot as plt
import numpy as np

# Intervalo de amostragem
n = np.arange(0, 101)

# Definição do número complexo
z = 0.95 * np.exp(1j * 0.2 * np.pi)

x_n = z ** n

# Criando figura com 4 subplots (2x2)
fig, axs = plt.subplots(2, 2, figsize=(12, 8))

# 1. Parte Real
axs[0, 0].stem(n, x_n.real)
axs[0, 0].set_title('Parte Real: Re{x[n]}')
axs[0, 0].set_xlabel('n')
axs[0, 0].set_ylabel('Amplitudes')
axs[0, 0].grid(True)

# 2. Parte Imaginária
axs[0, 1].stem(n, x_n.imag)
axs[0, 1].set_title('Parte Imaginária: Im{x[n]}')
axs[0, 1].set_xlabel('n')
axs[0, 1].set_ylabel('Amplitudes')
axs[0, 1].grid(True)

# 3. Módulo
axs[1, 0].stem(n, np.abs(x_n))
axs[1, 0].set_title('Módulo: |x[n]|')
axs[1, 0].set_xlabel('n')
axs[1, 0].set_ylabel('Módulo')
axs[1, 0].grid(True)

# 4. Trajetória no Plano Complexo
axs[1, 1].plot(x_n.real, x_n.imag, 'o-', markersize=3, linewidth=1)
axs[1, 1].axhline(0, color='gray')
axs[1, 1].axvline(0, color='gray')
axs[1, 1].set_aspect('equal')
axs[1, 1].set_title('Trajetória no Plano Complexo')
axs[1, 1].set_xlabel('Re{x[n]}')
axs[1, 1].set_ylabel('Im{x[n]}')
axs[1, 1].grid(True)

plt.tight_layout()
plt.show()
```
