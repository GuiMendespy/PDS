```python
import matplotlib.pyplot as plt
import numpy as np

# Índices discretos de n = 0 até 30
n = np.arange(0, 31)

# Sinais em tempo discreto
x1 = np.cos(0.3 * np.pi * n)
x2 = np.cos(2.3 * np.pi * n)

# Plotagem
plt.figure(figsize=(10, 4))
plt.stem(n, x1, linefmt='b-', markerfmt='bo', basefmt='r-', label='x1[n] = cos(0.3*pi*n)')
plt.stem(n, x2, linefmt='r--', markerfmt='rx', basefmt='r-', label='x2[n] = cos(2.3*pi*n)')

plt.title('Comparação de x1[n] e x2[n]')
plt.xlabel('n (amostras)')
plt.ylabel('Amplitudes')
plt.grid(True)
plt.legend()

plt.show()
```
