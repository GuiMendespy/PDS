```python
import matplotlib.pyplot as plt
import numpy as np

def z(theta):
    return np.exp(1j * theta)

theta = np.linspace(0, 2 * np.pi, 500)
valores = z(theta)

# Criando figura com 3 subplots
fig, axs = plt.subplots(1, 3, figsize=(15, 4))

# 1. Re{z(theta)} em função de theta
axs[0].plot(theta, valores.real)
axs[0].set_title('Re{z(theta)} vs theta')
axs[0].set_xlabel('theta (rad)')
axs[0].set_ylabel('Re{z}')
axs[0].grid(True)

# 2. Im{z(theta)} em função de theta
axs[1].plot(theta, valores.imag)
axs[1].set_title('Im{z(theta)} vs theta')
axs[1].set_xlabel('theta (rad)')
axs[1].set_ylabel('Im{z}')
axs[1].grid(True)

# 3. Im{z} versus Re{z} (Plano Complexo)
axs[2].plot(valores.real, valores.imag)
axs[2].axhline(0, color='gray')
axs[2].axvline(0, color='gray')
axs[2].set_aspect('equal')
axs[2].set_title('Im{z} vs Re{z}')
axs[2].set_xlabel('Re{z}')
axs[2].set_ylabel('Im{z}')
axs[2].grid(True)

plt.tight_layout()
plt.show()
```

e^(j*theta), cos(theta) e sin(theta) estão relacionados pela Fórmula de Euler, onde o cosseno representa a parte real de e^(j*theta) e o seno representa a parte imaginária.
