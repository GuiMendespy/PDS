```python
import matplotlib.pyplot as plt
import numpy as np

pontos = [
    {"label": "a) $e^{j0} = 1$", "val": 1 + 0j, "color": "red", "pos": (1.1, 0.1)},
    {"label": "b) $e^{j\pi/2} = j$", "val": 0 + 1j, "color": "blue", "pos": (0.1, 1.1)},
    {"label": "c) $e^{j\pi} = -1$", "val": -1 + 0j, "color": "green", "pos": (-1.7, 0.1)},
    {"label": "d) $e^{j3\pi/2} = -j$", "val": 0 - 1j, "color": "purple", "pos": (0.1, -1.2)},
    {"label": "e) $e^{j2\pi} = 1$", "val": 1 + 0j, "color": "orange", "pos": (1.1, -0.2)},
]

fig, ax = plt.subplots(figsize=(6, 6))

# Circunferência de raio unitário
theta = np.linspace(0, 2 * np.pi, 100)
ax.plot(np.cos(theta), np.sin(theta), '--', color='gray', alpha=0.5)

# Eixos e Origem
ax.axhline(0, color='black', linewidth=1.2)
ax.axvline(0, color='black', linewidth=1.2)
ax.plot(0, 0, 'ko', markersize=5)

# Vetores e identificações
for p in pontos:
    z = p["val"]
    ax.quiver(0, 0, z.real, z.imag, angles='xy', scale_units='xy', scale=1, color=p["color"], width=0.007)
    ax.plot(z.real, z.imag, 'o', color=p["color"])
    ax.text(p["pos"][0], p["pos"][1], p["label"], fontsize=10, color=p["color"], fontweight='bold')

ax.grid(True, linestyle=':', alpha=0.6)
ax.set_xlim(-1.8, 1.8)
ax.set_ylim(-1.8, 1.8)
ax.set_xlabel('Eixo Real (Re)')
ax.set_ylabel('Eixo Imaginário (Im)')
ax.set_title('Representação dos Resultados no Plano Complexo')
ax.set_aspect('equal', adjustable='box')

plt.show()
```
