```python
# Definindo os números complexos e posições dos textos
vetores = [
    {"label": "$z_1 = 1 + j$", "val": 1 + 1j, "color": "red", "pos": (1.1, 0.7)},
    {"label": "$z_2 = 2 + 2j$", "val": 2 + 2j, "color": "green", "pos": (2.1, 2.1)},
    {"label": "$z_3 = -2 + j$", "val": -2 + 1j, "color": "blue", "pos": (-2.4, 1.2)},
    {"label": "$z_4 = -1 - 2j$", "val": -1 - 2j, "color": "magenta", "pos": (-1.1, -2.4)},
]

fig, ax = plt.subplots(figsize=(7, 7))

# 1. Eixo Real e Eixo Imaginário
ax.axhline(0, color='black', linewidth=1.2)
ax.axvline(0, color='black', linewidth=1.2)

# 2. Vetores apontando a partir da origem
for v in vetores:
    z = v["val"]
    ax.quiver(0, 0, z.real, z.imag, angles='xy', scale_units='xy', scale=1, 
              color=v["color"], width=0.008)
    ax.text(v["pos"][0], v["pos"][1], v["label"], fontsize=11, color=v["color"], fontweight='bold')

# 3. Origem
ax.plot(0, 0, 'ko', markersize=6)
ax.text(0.15, -0.25, 'Origem (0,0)', fontsize=10)

# 4. Grade, limites e rótulos
ax.grid(True, linestyle='--', alpha=0.6)
ax.set_xlim(-3.5, 3.5)
ax.set_ylim(-3.5, 3.5)
ax.set_xlabel('Eixo Real (Re)', fontsize=12)
ax.set_ylabel('Eixo Imaginário (Im)', fontsize=12)
ax.set_title('Plano Complexo', fontsize=14)
ax.set_aspect('equal', adjustable='box')

plt.show()
```
