```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Definição da grade de frequências no intervalo [0, pi] com 1000 pontos
w = np.linspace(0, np.pi, 1000)

# 2. Cálculo numérico de H(e^(j*w))
H = 1 / (1 - 0.8 * np.exp(-1j * w))

# 3. Extração da magnitude e da fase em radianos
magnitude = np.abs(H)
phase = np.angle(H)

# 4. Cálculo e identificação dos pontos notáveis (w = 0, pi/2, pi)
w_points = np.array([0, np.pi / 2, np.pi])
w_labels = [r"$0$", r"$\pi/2$", r"$\pi$"]
H_points = 1 / (1 - 0.8 * np.exp(-1j * w_points))
mag_points = np.abs(H_points)
phase_points = np.angle(H_points)

# 5. Construção dos gráficos
plt.figure(figsize=(10, 8))

# --- Gráfico de Magnitude ---
plt.subplot(2, 1, 1)
plt.plot(w, magnitude, "b-", linewidth=2, label=r"$|H(e^{j\omega})|$")
plt.scatter(w_points, mag_points, color="red", zorder=5)

for i in range(len(w_points)):
    plt.annotate(
        f"({w_labels[i]}, {mag_points[i]:.4f})",
        (w_points[i], mag_points[i]),
        textcoords="offset points",
        xytext=(15, -12 if i == 0 else 10),
        ha="center",
        fontsize=9,
        bbox=dict(boxstyle="round,pad=0.3", fc="lightyellow", ec="orange"),
    )

plt.title(
    r"Resposta em Frequência $H(e^{j\omega}) = \frac{1}{1 - 0.8e^{-j\omega}}$",
    fontsize=12,
)
plt.ylabel(r"Magnitude $|H(e^{j\omega})|$", fontsize=11)
plt.xticks(
    [0, np.pi / 4, np.pi / 2, 3 * np.pi / 4, np.pi],
    [r"$0$", r"$\pi/4$", r"$\pi/2$", r"$3\pi/4$", r"$\pi$"],
)
plt.grid(True, linestyle="--", alpha=0.7)
plt.legend(loc="upper right")

# --- Gráfico de Fase ---
plt.subplot(2, 1, 2)
plt.plot(w, phase, "r-", linewidth=2, label=r"$\angle H(e^{j\omega})$")
plt.scatter(w_points, phase_points, color="blue", zorder=5)

for i in range(len(w_points)):
    plt.annotate(
        f"({w_labels[i]}, {phase_points[i]:.4f} rad)",
        (w_points[i], phase_points[i]),
        textcoords="offset points",
        xytext=(20, 10 if i != 1 else -18),
        ha="center",
        fontsize=9,
        bbox=dict(boxstyle="round,pad=0.3", fc="lightcyan", ec="blue"),
    )

plt.xlabel(r"Frequência Angular $\omega$ (rad/amostra)", fontsize=11)
plt.ylabel(r"Fase $\angle H(e^{j\omega})$ (rad)", fontsize=11)
plt.xticks(
    [0, np.pi / 4, np.pi / 2, 3 * np.pi / 4, np.pi],
    [r"$0$", r"$\pi/4$", r"$\pi/2$", r"$3\pi/4$", r"$\pi$"],
)
plt.grid(True, linestyle="--", alpha=0.7)
plt.legend(loc="upper right")

plt.tight_layout()
plt.show()
```
