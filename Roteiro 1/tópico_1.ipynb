```python
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros do sinal
f0 = 5        # Frequência do sinal: 5 Hz
A = 1.0       # Amplitude
duration = 1  # Duração de 1 segundo

# 1. Sinal Contínuo (Simulado com alta taxa de pontos)
t_cont = np.linspace(0, duration, 1000)
x_cont = A * np.sin(2 * np.pi * f0 * t_cont)

# 2. Sinal Discreto (Amostrado a fs = 30 Hz)
fs = 30       # Taxa de amostragem
Ts = 1 / fs   # Período de amostragem
n = np.arange(0, int(fs * duration))
t_disc = n * Ts
x_disc = A * np.sin(2 * np.pi * f0 * t_disc)

# Gerando os gráficos
plt.figure(figsize=(12, 5))

# Plot do Sinal Contínuo
plt.subplot(1, 2, 1)
plt.plot(t_cont, x_cont, 'b-', linewidth=2, label=r'$x(t) = \sin(2\pi \cdot 5 \cdot t)$')
plt.title('Sinal Contínuo no Tempo $x(t)$')
plt.xlabel('Tempo $t$ (segundos)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.legend()

# Plot do Sinal Discreto
plt.subplot(1, 2, 2)
plt.stem(n, x_disc, linefmt='r-', markerfmt='ro', basefmt='k-', label=r'$x[n] = \sin(2\pi \cdot 5 \cdot n T_s)$')
plt.title('Sinal Discreto no Tempo $x[n]$')
plt.xlabel('Índice da Amostra $n$')
plt.ylabel('Amplitude')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
